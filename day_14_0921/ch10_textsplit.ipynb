### 텍스트 분할

```python
# data/appendix-keywords.txt 파일을 열어서 f 파일 객체 생성
with open(r"langchain-kr\07-TextSplitter\data\appendix-keywords.txt", encoding="utf-8") as f:
    file = f.read() # 파일의 내용을 읽어서 file 변수에 저장
```

```python
print(file[:500])
```

```python
from langchain_text_splitters import CharacterTextSplitter

text_splitter = CharacterTextSplitter(
    separator="\n\n", # 텍스트를 분할할 구분자
    chunk_size=210, # 청크의 최대 크기(문자 수)
    chunk_overlap=0, # 청크 간의 중복되는 문자 수
    length_function=len, # 텍스트의 길이 계산
)
```

```python
texts = text_splitter.create_documents([file]) # 텍스트를 청크로 분할
print(len(texts[0].page_content)) # 글자 개수 출력
print(texts[0]) # 분할된 문서 중 첫 번째 문서 출력
```

```python
metadatas = [
    {"document": 1},
    {"document": 2},
]
documents = text_splitter.create_documents(
    [
        file,
        file,
    ],
    metadatas=metadatas,
)
print(documents[0])
```

### Semantic Chunker의 문장 분할 기준점 설정하기

```python
# 백분위수 (percentile) 분리 방식
text_splitter = SemanticChunker(
    OpenAIEmbeddings(),
    breakpoint_threshold_type="percentile",
    breakpoint_threshold_amount=70,
)

docs = text_splitter.create_documents([file])
for i, doc in enumerate(docs[:5]):
    print(f"[Chunk {i}]", end="\n\n")
    print(doc.page_content)
    print("===" * 20)

print(len(docs))
```

```python
# 표준편차 (standard deviation) 분리 방식
text_splitter = SemanticChunker(
    OpenAIEmbeddings(),
    breakpoint_threshold_type="standard_deviation",
    breakpoint_threshold_amount=1.25,
)

docs = text_splitter.create_documents([file])
for i, doc in enumerate(docs[:5]):
    print(f"[Chunk {i}]", end="\n\n")
    print(doc.page_content)
    print("===" * 20)

print(len(docs))
```

```python
# 사분위수 범위 (interquartile range) 분리 방식
text_splitter = SemanticChunker(
    OpenAIEmbeddings(),
    breakpoint_threshold_type="interquartile",
    breakpoint_threshold_amount=0.5,
)

docs = text_splitter.create_documents([file])

for i, doc in enumerate(docs[:5]):
    print(f"[Chunk {i}]", end="\n\n")
    print(doc.page_content)
    print("===" * 20)

print(len(docs))
```

```python
text_splitter = SemanticChunker(
    OpenAIEmbeddings(),
    breakpoint_threshold_type="interquartile",
    breakpoint_threshold_amount=0.5,
)

docs = text_splitter.create_documents([file])

for i, doc in enumerate(docs[:5]):
    print(f"[Chunk {i}]")
    print(doc.page_content)
    print("===" * 20, end="\n\n")

print(len(docs))
```

### 코드 분할하기

```python
from langchain_text_splitters import(
    Language, RecursiveCharacterTextSplitter,
)

PYTHON_CODE = """
def hello_world():
    print("Hello, World")
    
hello_world()
"""

python_splitter = RecursiveCharacterTextSplitter.from_language(
    language=Language.PYTHON, chunk_size=50, chunk_overlap=0
)

python_docs = python_splitter.create_documents([PYTHON_CODE])

for doc in python_docs:
    print(doc.page_content, end="\n============\n")
```

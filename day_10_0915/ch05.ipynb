```python
template = "What is the population of {country}"

# from_template 메서드를 이용하여 PromptTemplate 객체 생성
prompt = PromptTemplate.from_template(template)

chain = prompt | llm    # chain 생성

# country 변수에 입력된 값이 자동으로 치환되어 수행됨
chain.invoke("The United States").content
```

```python
template = "What is the population of {country}?"

# PromptTemplate 객체를 활용하여 prompt_template 생성
prompt = PromptTemplate(
    template=template,
    input_variables=["country"],
)

prompt
```

```python
prompt.format(country="The United States")
```

```python
template = "What is the population of {country1} and {country2}?"

# PromptTemplate 객체를 활용하여 prompt_template 생성
prompt = PromptTemplate(
    template=template,
    input_variables=["country1"],
    partial_variables={
        "country2": "Mexico"
    },
)

prompt
```

```python
prompt.format(country1="The United States")
```

```python
prompt_partial = prompt.partial(country2="Canada")
prompt_partial
```

```python
prompt_partial.format(country1="The United States")
```

```python
chain = prompt_partial | llm

chain.invoke("The United States").content
```

```python
# 덮어쓰기
chain.invoke({"country1": "The United States", "country2": "Japan"}).content
```

### 부분 변수 활용하기

```python
from datetime import datetime

datetime.now().strftime("%B %d")    #오늘 날짜 출력
```

```python
def get_today():    # 날짜를 반환하는 함수 정의
    return datetime.now().strftime("%B %d")
```

```python
prompt = PromptTemplate(
    template="Today is {today}. Tell me {n} celebrities who died on this day.",
    input_variables=["n"],
    partial_variables={
        "today": get_today  # dictionary 형태로 partial_variables를 전달
    },
)
```

```python
prompt.format(n=2)
```

```python
chain = prompt | llm
print(chain.invoke(2).content)
```

```python
print(chain.invoke({"today": "Jan 17", "n": 2}).content)
```

```python
from langchain_core.prompts import ChatPromptTemplate

chat_template = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a kind AI assistant, and your name is Tim."),
        ("human", "Hello!"),
        ("ai", "Hello! How can I help you?"),
        ("human", "{user_input}"),
    ]
)
```

```python
messages = chat_template.format_messages(
    name = "Tim", user_input = "What is your name?"
)
messages
```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI()
llm.invoke(messages).content
```

```python
chain = chat_template | llm

chain.invoke({"name": "Tim", "user_input": "What is your name?"}).content
```

### MessagesPlaceholder

```python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

chat_prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system", "You are an AI assistant specifically for summarizing. Your task is to summarize the chat with keywords.",
        ),
        MessagesPlaceholder(variable_name="conversation"),
        ("human", "Summarize the chat with {word_count} keywords."),
    ]
)
chat_prompt
```

```python
formatted_chat_prompt = chat_prompt.format(
    word_count=5,
    conversation=[
        ("human", "Hello! I'm Jacob, and I just joined the company today. Nice to meet you!"),
        ("ai", "Hello! It is a pleasure to meet you!"),
    ],
)

print(formatted_chat_prompt)
```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI()

chain = chat_prompt | llm | StrOutputParser()
```

```python
chain.invoke(
    {
        "word_count": 5,
        "conversation": [
            (
                "human", "Hello! I'm Jacob, and I just joined the company today. Nice to meet you!",
            ),
            ("ai", "Hello! It is a pleasure to meet you!"),
        ],
    }
)
```

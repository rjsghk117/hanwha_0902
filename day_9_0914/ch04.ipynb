```python
from langchain_openai import ChatOpenAI
```

```python
# Chat Model 생성
llm = ChatOpenAI(
    temperature=0.1,
    model="gpt-4o-mini",
)
```

```python
#응답
response = llm.invoke(question)
```

### logprob 로그 프로버블리티 활성화
- 모델의 답변 확신도(Confidence level) :0에 가까울수록 정답에 가까운 확신을 가진다.

```python
# Object 생성
llm_with_logprob = ChatOpenAI(
    temperature=0.1,
    max_tokens=2048,
    model="gpt-4o-mini"
).bind(logprobs=True)
```

```python
question2 = "What is the capitla city of India?"

response = llm_with_logprob.invoke(question2)
```

```python
response
```

```python
response.response_metadata
```

```python
response.response_metadata["logprobs"]
```

### Stream 실시간 출력 방식

```python
#답변을 한 토큰씩 받아와서 화면에 실시간으로 뿌려줄 때
answer = llm.stream("10 best workouts for beginners")
```

```python
#flush=False: 결과가 버퍼에 저장되었다가, 한번에 출력
#flush=True: 결과가 생성되는대로 바로바로 출력

for token in answer:
    print(token.content, end="", flush=True)
```

```python
#Streaming 답변겨로가 변수에 저장

answer = llm.stream("10 best workouts for beginners")

final_answer = ""
for token in answer:
    final_answer += token.content
```
